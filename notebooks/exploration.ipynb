# AdamW Demo Exploration

Load training metrics and visualize loss, accuracy, and learning-rate behavior.

```python
from pathlib import Path
import json
import matplotlib.pyplot as plt

history_path = Path('../artifacts/history.json')
if not history_path.exists():
    raise FileNotFoundError('Run training first: python src/main.py')

with history_path.open('r', encoding='utf-8') as f:
    history = json.load(f)

epochs = range(1, len(history['train_loss']) + 1)
history.keys()
```

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

axes[0].plot(epochs, history['train_loss'], label='train_loss')
axes[0].plot(epochs, history['val_loss'], label='val_loss')
axes[0].set_title('Loss')
axes[0].set_xlabel('Epoch')
axes[0].grid(alpha=0.3)
axes[0].legend()

axes[1].plot(epochs, history['train_acc'], label='train_acc')
axes[1].plot(epochs, history['val_acc'], label='val_acc')
axes[1].set_title('Accuracy')
axes[1].set_xlabel('Epoch')
axes[1].set_ylim(0.0, 1.0)
axes[1].grid(alpha=0.3)
axes[1].legend()

axes[2].plot(epochs, history['lr'], label='learning_rate')
axes[2].set_title('Learning Rate')
axes[2].set_xlabel('Epoch')
axes[2].grid(alpha=0.3)
axes[2].legend()

plt.tight_layout()
plt.show()
```
